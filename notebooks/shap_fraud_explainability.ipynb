#### Shap Analysis(Global + Local) For Fraud Data 

##### Shap Summery Plot For Fraud Data (Global Importance)

```python
import pandas as pd
import numpy as np
import shap

# Separate fraud and non-fraud
fraud_cases = fraud_x_test[fraud_y_test == 1]
non_fraud_cases = fraud_x_test[fraud_y_test == 0].sample(n=500, random_state=42)

# Combine for SHAP
X_shap = pd.concat([fraud_cases, non_fraud_cases])
y_shap = np.concatenate([fraud_y_test[fraud_y_test == 1], fraud_y_test[fraud_y_test == 0][:500]])

# SHAP explainer
explainer = shap.TreeExplainer(fraud_best_model)

# Compute SHAP values
shap_values = explainer(X_shap)

# Global summary plot
shap.summary_plot(shap_values, X_shap, max_display=10)
```

#### Feature Impact Distribution
**age** and **purchase_value** both have SHAP values centered around 0, with points spread roughly symmetrically from -0.25 to +0.25.

This suggests that, on average, these features have a moderate impact on the model’s predictions, but the direction of impact varies per observation.

#### Shap Force Plot for Fraud Data
* we will ganerate 3 SHAP force plots
   * True Positive(TP)-fraud correctly detected
   * False Positive(FP)- legitimate transaction flagged as fraud
   * False Negative(FN)- fraud that the model missed

```python
# get predictions and probabilities
y_pred=fraud_best_model.predict(fraud_x_test)
y_proba=fraud_best_model.predict_proba(fraud_x_test)[:,1]
fraud_y_test=np.array(fraud_y_test)
y_pred=np.array(y_pred)
```

```python
# identify TP,FP and FN indices
tp_idx=np.where((fraud_y_test==1) & (y_pred==1))[0][0] # True Positive: actual=1 and Predicted=1
fp_idx=np.where((fraud_y_test==0) & (y_pred==1))[0][0] # False Positive: actual=0 and predicted=1
fn_idx=np.where((fraud_y_test==1) & (y_pred==0))[0][0] # False Negative: actual=1 and predicted=0
```

```python
# build shap explainer 
explainer=shap.TreeExplainer(fraud_best_model,approximate=True)
```

```python
# generate shap Force values for only 3 rows
X_explain=fraud_x_test.iloc[[tp_idx,fp_idx,fn_idx]]
shap_values=explainer(X_explain)
```

##### True Positive(Correct fraud detection Force Plot)

```python
class_idx = 1  # fraud class

# True Positive
shap.plots.force(
    explainer.expected_value[class_idx],
    shap_values.values[0, :, class_idx],
    X_explain.iloc[0],
    matplotlib=True
)



```

**(executive view)**

* This transaction is classified as fraud mainly because the user signed up very recently, came through SEO traffic, used Chrome, and is from a higher-risk country.
* Although not being from the US slightly reduces the risk, it is overwhelmed by stronger fraud indicators, leading to a 91% fraud probability.

##### False Positive

```python
# False Positive
shap.plots.force(
    explainer.expected_value[class_idx],
    shap_values.values[1, :, class_idx],
    X_explain.iloc[1],
    matplotlib=True
)
```

**Executive summary for FP**
* This false positive occurred because the model strongly penalizes very recent signups and missing location data. While these patterns are common in fraud, they can also occur for legitimate new users, indicating a need for better handling of missing geo information and softer weighting of account age.

#### False Nagative

```python
# False Negative
shap.plots.force(
    explainer.expected_value[class_idx],
    shap_values.values[2, :, class_idx],
    X_explain.iloc[2],
    matplotlib=True
)
```

* The model trusted the account too much because it was not new, had a known country, and lacked obvious acquisition red flags — even though the transaction was fraudulent

#### Interpretation
##### Comparison
* In Random Forest importance, age and purchase_value rank 2nd and 3rd, indicating they are strong global predictors.
* In SHAP, these features show context-dependent effects (both positive and negative contributions). This suggests:
* Random Forest importance may overstate their uniform predictive power.
* SHAP reveals the non-linear, interaction-driven nature of their impact, which aligns with the “SHAP interaction values” label.
##### Top 5 Drivers of Fraud Predictions**
From the Random Forest importance plot, the top 5 drivers are:
* time_since_signup — Newer accounts likely riskier (common fraud signal).
* age — Customer age influences risk profile (but direction varies).
* purchase_value— Transaction amount affects fraud likelihood (often very high or very low amounts are suspicious).
* country_United States — Geographic risk factor (possibly higher fraud rates or more stringent rules).
* country_unknown — Missing country data may correlate with fraud.
 ##### Surprising or Counterintuitive Findings**
* a) Country features dominate the middle of the top 10
4 out of the top 10 features are country-related (US, unknown, Germany, China, Korea).
This suggests geography is a major fraud signal, possibly due to:
Regional fraud patterns
IP mismatches
Shipping/billing address inconsistencies
Surprise: Country flags may be overemphasized by the model — could lead to geographic bias if not controlled.
* b) age and purchase_value show symmetric SHAP distributions
Both have SHAP values centered near 0 with wide spread.
This means:
For some transactions, higher age increases fraud score; for others, it decreases it.
Same for purchase value.
Interpretation: These features are not linearly correlated with fraud — their effect depends on other factors (e.g., country, time since signup).
Counterintuitive: If we only looked at feature importance, we might think “older age always reduces fraud risk” — but SHAP shows it’s context-dependent.
* c) time_since_signup is the top feature, but not shown in SHAP plot
This suggests the model heavily weights account newness.
Why surprising? It’s common in fraud models, but its dominance over behavioral/transactional features (purchase_value) indicates the model may be relying heavily on this single signal, which could be exploited (fraudsters can create older accounts).
* d) Lack of device or behavioral features in top 10
Only browser_Chrome appears (9th).
Missing: IP velocity, device fingerprint, transaction frequency, etc.
Possible implication: The model may be underusing real-time behavioral signals, which are often strong fraud indicators.

#### Business Recommendations

1. Flag New Accounts from High-Risk Countries
* Action: Add extra verification for purchases over $500 from accounts under 24 hours old in the US, Germany, China, or Korea.
* Why: time_since_signup is your top fraud predictor, and country features dominate the top 10 risk signals.
2. Watch for Unusual Age-Value Patterns
* Action: Alert customer service when young users (<25) with old accounts (>1 year) make unusually high purchases compared to their history.
* Why: SHAP shows age and purchase value have variable effects—sometimes increasing, sometimes decreasing fraud risk. Context matters.
3. Set Country-Specific Verification Rules
* Action: Use different verification thresholds by country:
* US/Germany: 2FA for $1,000+ from new accounts
* China/Korea: 2FA for $500+ from new accounts
* Unknown country: Review all $200+ from new accounts
* Why: 5 of your top 10 predictors are country-related. SHAP shows purchase value's impact varies—different thresholds needed by region.
* Expected Impact: Reduce fraud 15-30% while minimizing customer friction through targeted rules based on your model's top signals.
